```python

import re
import nltk
import numpy as np
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('averaged_perceptron_tagger')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from collections import Counter
```

```python

text = """Artificial intelligence is my favorite technology because it makes computers smarter.
AI helps students learn concepts and solve difficult problems.
Machine learning allows computers to discover patterns from data.
I enjoy exploring AI because it is useful in healthcare, education, and transportation.
In the future, intelligent systems may make our daily lives easier and more productive."""

clean_text = re.sub(r'[^\w\s]', '', text.lower())

print(clean_text)
```

```python

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("Word tokens:", words)
print("\nSentences:", sentences)
print("\nNumber of words:", len(words))
print("Number of sentences:", len(sentences))
```

```python

sample = "AI is useful, isn't it? Machine-learning is interesting!"

print("Using split():")
print(sample.split())

print("\nUsing word_tokenize():")
print(word_tokenize(sample))
```

```python

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("Words after stopword removal:")
print(filtered_words)
```

```python
#Q2.

alpha_words = re.findall(r'\b[A-Za-z]+\b', text.lower())

print("Alphabetic words:")
print(alpha_words)
```

```python

filtered_alpha = [
    word for word in alpha_words
    if word not in stop_words
]

print(filtered_alpha)

```

```python

porter = PorterStemmer()

stemmed_words = [
    porter.stem(word)
    for word in filtered_alpha
]

print("Stemmed words:")
print(stemmed_words)
```

```python

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_alpha
]

print("Lemmatized words:")
print(lemmatized_words)
```

```python

comparison = list(zip(
    filtered_alpha,
    stemmed_words,
    lemmatized_words
))

print("Original Word | Stem | Lemma")
for original, stem, lemma in comparison:
    if stem != lemma:
        print(original, "|", stem, "|", lemma)
```

```python
#Q3.

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

documents = [
    "Artificial intelligence helps computers learn and make decisions.",
    "Blockchain stores secure transactions in a distributed ledger.",
    "Machine learning uses data to predict future results."
]

# Bag of Words
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(documents)

print("Bag of Words:")
print("Features:", count_vectorizer.get_feature_names_out())
print(bow_matrix.toarray())
```

```python

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

features = tfidf_vectorizer.get_feature_names_out()

print("TF-IDF Matrix:")
print(np.round(tfidf_matrix.toarray(), 3))

# Top 3 keywords for each document
for i, doc in enumerate(documents):
    scores = tfidf_matrix[i].toarray().flatten()
    top_indices = scores.argsort()[::-1][:3]

    print(f"\nDocument {i+1}: {doc}")
    for index in top_indices:
        print(features[index], ":", round(scores[index], 3))
```

```python
#Q4.

from sklearn.metrics.pairwise import cosine_similarity

ai_text = """
Artificial intelligence enables machines to learn from data.
AI systems recognize patterns and support decision making.
Machine learning improves predictions using experience.
AI is useful in healthcare, education, and robotics.
Intelligent systems can automate repetitive tasks.
"""

blockchain_text = """
Blockchain records transactions across a distributed network.
Each block contains information linked to earlier blocks.
Cryptography helps protect blockchain records.
Blockchain supports secure payments and digital contracts.
The network maintains a shared transaction history.
"""

# Preprocessing and tokenization
def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    tokens = word_tokenize(text)
    return [word for word in tokens if word not in stop_words]

tokens_ai = preprocess(ai_text)
tokens_blockchain = preprocess(blockchain_text)

print("AI tokens:", tokens_ai)
print("Blockchain tokens:", tokens_blockchain)
```

```python

set_ai = set(tokens_ai)
set_blockchain = set(tokens_blockchain)

intersection = set_ai & set_blockchain
union = set_ai | set_blockchain

jaccard = len(intersection) / len(union) if union else 0

print("Common words:", intersection)
print("Jaccard Similarity:", round(jaccard, 4))
```

```python

texts = [ai_text, blockchain_text]

vectorizer = TfidfVectorizer(stop_words='english')
tfidf = vectorizer.fit_transform(texts)

cosine = cosine_similarity(tfidf[0], tfidf[1])[0][0]

print("Cosine Similarity:", round(cosine, 4))
```
